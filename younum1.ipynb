```python
import numpy as np        # import numpy

```

```python
# 1 dimensional array
list1=[1,2,3,4,3]
arr1=np.array(list1)
arr1
print(arr1.ndim)   #dimensional
```

```python
print(list1*2)
print(arr1*2)
type(arr1) #dtype
arr1.dtype   # datatype
```

```python
arr2=np.array([1,2,3,4])    #another form
print(arr2)
arr2.size            #to find the array  size
arr2.shape           #find the shape(how many columns and rows)
arr2.dtype            #datatype
```

```python
# 2 dimensional array
arr21=np.array([
    [1,2,3],
    [4,3,2],
    [1,4,9],
])
print(arr21)
type(arr21)         #to find the array type(dtype)
arr21.size        #to find the array  size
arr21.ndim        #to find the dimensional
arr21.shape       ##find the shape(how many columns and rows)
arr21.dtype       #datatype

```
